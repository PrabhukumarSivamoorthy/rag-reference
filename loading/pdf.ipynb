# PDF Loading

Load a PDF into documents/text, ready for chunking. Common loaders: `pypdf` / `PyPDFLoader`, `pdfplumber`, `PyMuPDF (fitz)`, or `UnstructuredPDFLoader` for richer layout extraction.

```python
# Sample input lives in the repo:
sample = ROOT / "assets/sample-docs/sdlc-end-to-end.pdf"
print(sample, sample.exists())
```

```python
# Load the PDF with LangChain's PyPDFLoader (backed by pypdf).
from langchain_community.document_loaders import PyPDFLoader

# PyPDFLoader splits the PDF into one Document per page,
# each with .page_content (extracted text) and .metadata (source, page number).
loader = PyPDFLoader(str(sample))
data = loader.load()

# Sanity check: page count, first page's metadata, and a short text preview.
print(f"Loaded {len(data)} pages")
print(data[0].metadata)
print(data[0].page_content[:500])
```
