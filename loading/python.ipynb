# Python (.py) Loading

`PythonLoader` reads a `.py` source file into one `Document` (raw source as text). For code-aware splitting that respects functions/classes, see `RecursiveCharacterTextSplitter.from_language(Language.PYTHON)` in chunking.

```python
from langchain_community.document_loaders import PythonLoader

# Use a real .py file from the repo as the sample.
sample = ROOT / "env_checker.py"
docs = PythonLoader(sample).load()

# Sanity check: document count, metadata, and a short source preview.
print(f"Loaded {len(docs)} document(s)")
print(docs[0].metadata)
print(docs[0].page_content[:500])
```
