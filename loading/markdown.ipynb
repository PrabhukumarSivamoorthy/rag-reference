# Markdown Loading

A `.md` file is plain text, so `TextLoader` reads it into one `Document`. `UnstructuredMarkdownLoader` parses structure but needs the `unstructured` package; for splitting by heading see `MarkdownHeaderTextSplitter` in chunking.

```python
from langchain_community.document_loaders import TextLoader

# Use a real markdown file from the repo as the sample.
sample = ROOT / "README.md"
docs = TextLoader(sample).load()

# Sanity check: document count, metadata, and a short text preview.
print(f"Loaded {len(docs)} document(s)")
print(docs[0].metadata)
print(docs[0].page_content[:500])
```
