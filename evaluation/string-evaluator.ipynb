# String Evaluator

LangChain **string evaluators** compare a `prediction` against a `reference` string and return a score — handy for checking a generated answer against a known-good one. In LangChain 1.x these live in the `langchain_classic.evaluation` module.

Two flavors shown below:
- **`string_distance`** — character-level (Levenshtein) distance. No model, free, but purely lexical.
- **`embedding_distance`** — semantic distance via embeddings (catches paraphrases).

Both are *distance* scores: **lower = more similar**.

## Prediction vs. reference

```python
prediction = "LLD is the detailed internal design of each component."
reference  = "Low-Level Design provides the detailed internal design of every component."
```

## 1. `string_distance` (lexical, free)

```python
from langchain_classic.evaluation import load_evaluator

string_eval = load_evaluator("string_distance")
print(string_eval.evaluate_strings(prediction=prediction, reference=reference))
```

## 2. `embedding_distance` (semantic)

```python
from langchain_openai import OpenAIEmbeddings

embed_eval = load_evaluator(
    "embedding_distance",
    embeddings=OpenAIEmbeddings(model="text-embedding-3-small"),
)
print(embed_eval.evaluate_strings(prediction=prediction, reference=reference))
```
